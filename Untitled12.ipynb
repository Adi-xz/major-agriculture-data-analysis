```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')
print("Shape of data:", df.shape)
print("\nMissing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0])
df.head()
```

```python
cols_to_fill = ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']
for col in cols_to_fill:
    df[col] = df.groupby('Season')[col].transform(lambda x: x.fillna(x.median()))
print("Missing values after cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
seasonal_summary = df.groupby('Season')[['Yield_Tonnes_Ha', 'Total_Cost_INR', 'Profit_INR', 'Water_Efficiency_t_per_1000m3']].mean()
pd.options.display.float_format = '{:,.2f}'.format
print("Average Performance by Season:")
print(seasonal_summary)
```

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=df, x='Season', y='Profit_INR', hue='Season', palette='viridis', ax=axes[0], errorbar=None, legend=False)
axes[0].set_title('Average Profit by Season')
axes[0].set_ylabel('Profit (INR)')
sns.scatterplot(data=df, x='Water_Efficiency_t_per_1000m3', y='Profit_INR', hue='Season', palette='viridis', alpha=0.6, ax=axes[1])
axes[1].set_title('Water Efficiency vs. Profit')
axes[1].axhline(0, color='red', linestyle='--', linewidth=1.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
key_metrics = ['Yield_Tonnes_Ha', 'Total_Cost_INR', 'Profit_INR',
               'Water_Efficiency_t_per_1000m3', 'Rainfall_mm', 'Fertilizer_kg_ha']
corr_matrix = df[key_metrics].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation of Key Agricultural Metrics')
plt.show()
```
